# Big 12 Ranking Comparison
Run the model first from the repository root with `python src/big12_ranking.py`.

```python
import pandas as pd
import matplotlib.pyplot as plt
ratings = pd.read_csv('results/Team_Ratings.csv')
games = pd.read_csv('results/Game_Residuals.csv')
```

```python
official_order = ['Texas Tech Red Raiders', 'BYU Cougars', 'Utah Utes', 'Houston Cougars', 'Arizona Wildcats', 'Arizona State Sun Devils', 'TCU Horned Frogs', 'Iowa State Cyclones', 'Cincinnati Bearcats', 'Kansas State Wildcats', 'Baylor Bears', 'Kansas Jayhawks', 'UCF Knights', 'West Virginia Mountaineers', 'Colorado Buffaloes', 'Oklahoma State Cowboys']
official_rank = {t:i for i,t in enumerate(official_order,1)}
comparison = ratings.copy()
comparison['Official_Rank'] = comparison['Team'].map(official_rank)
comparison['Rank_Difference'] = comparison['Rank'] - comparison['Official_Rank']
comparison.sort_values('Official_Rank')
```

```python
fig, ax = plt.subplots(figsize=(8,8))
ax.scatter(games['Real_Gap'], games['Predicted_Gap'], alpha=.75)
lo=min(games['Real_Gap'].min(),games['Predicted_Gap'].min())-5
hi=max(games['Real_Gap'].max(),games['Predicted_Gap'].max())+5
ax.plot([lo,hi],[lo,hi],'--')
ax.set_xlabel('Actual margin (Team A - Team B)'); ax.set_ylabel('Predicted margin')
ax.set_title('Actual vs Predicted Score Margins'); ax.grid(alpha=.25); plt.show()
```

The model fit metrics are in-sample because the 72 conference games were used to estimate the ratings.
